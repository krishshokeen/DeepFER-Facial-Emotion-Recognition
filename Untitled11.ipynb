# **Part 3 – Data preprocessing**

## **Data Preprocessing**
Preprocessing converts raw facial images into a consistent format suitable for deep learning.

The following operations are performed:
1. Convert images into RGB format.
2. Resize images to 224 × 224 pixels.
3. Convert images into PyTorch tensors.
4. Apply ImageNet normalisation.
5. Use augmentation on training images only.
ImageNet normalisation is used because the selected architectures are initialised with pretrained ImageNet weights

```python
# Define image transformations

train_transform = transforms.Compose([

    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),

    transforms.RandomHorizontalFlip(p=0.5),

    transforms.RandomRotation(degrees=10),

    transforms.ColorJitter(
        brightness=0.15,
        contrast=0.15
    ),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

evaluation_transform = transforms.Compose([

    transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),

    transforms.ToTensor(),

    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("Image transformations defined successfully.")
```

## **Training, Validation and Testing Split**
The original training dataset is divided into two subsets:
- Training set: used to update model parameters.
- Validation set: used to monitor model performance and select the best checkpoint.
The separate test dataset is retained for final performance evaluation.

A stratified split is used to preserve the approximate distribution of emotion categories across training and validation subsets

```python
# Create stratified training and validation indices

labels = np.array([
    label for _, label in raw_train_dataset.samples
])

indices = np.arange(len(labels))

train_indices, val_indices = train_test_split(
    indices,
    test_size=0.20,
    random_state=SEED,
    stratify=labels
)

print("Training samples:", len(train_indices))
print("Validation samples:", len(val_indices))
print("Testing samples:", len(raw_test_dataset))
```

```python
# Create separate datasets with different transforms

class FERDataset(Dataset):

    def __init__(self, samples, transform=None):
        self.samples = samples
        self.transform = transform

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, index):

        image_path, label = self.samples[index]

        image = Image.open(image_path).convert("RGB")

        if self.transform:
            image = self.transform(image)

        return image, label


train_samples = [
    raw_train_dataset.samples[i]
    for i in train_indices
]

val_samples = [
    raw_train_dataset.samples[i]
    for i in val_indices
]

train_dataset = FERDataset(
    train_samples,
    train_transform
)

val_dataset = FERDataset(
    val_samples,
    evaluation_transform
)

test_dataset = FERDataset(
    raw_test_dataset.samples,
    evaluation_transform
)

print("Datasets created successfully.")
```

## **Handling Class Imbalance**
Class imbalance occurs when some emotion categories contain significantly more images than others.

A model trained on an imbalanced dataset may become biased towards frequently occurring classes.

To reduce this problem, the project uses a Weighted Random Sampler. It assigns higher sampling probabilities to images belonging to underrepresented classes.

This allows minority emotion categories to receive more attention during training

```python
# Calculate class weights for the training subset

train_labels = np.array([
    label for _, label in train_samples
])

class_counts = np.bincount(
    train_labels,
    minlength=NUM_CLASSES
)

class_weights = 1.0 / np.maximum(class_counts, 1)

sample_weights = class_weights[train_labels]

sampler = WeightedRandomSampler(
    weights=torch.DoubleTensor(sample_weights),
    num_samples=len(sample_weights),
    replacement=True
)

print("Training class counts:", class_counts)
print("Class weights:", class_weights)
```

```python
# Create data loaders

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    sampler=sampler,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

print("Data loaders created successfully.")
```

# **Part 4 – Deep learning model development**

## **Model Development**
Three pretrained deep learning architectures are selected for comparative analysis.

1. ConvNeXt-Tiny

ConvNeXt is a modern convolutional neural network architecture that incorporates design improvements inspired by vision transformers.

2. ResNet18

ResNet18 uses residual connections to improve information flow through deep neural networks. It provides a relatively lightweight architecture for image classification.

3. Swin V2 Tiny

Swin Transformer V2 uses hierarchical transformer blocks and shifted-window attention to learn visual relationships at different scales.

All three architectures use pretrained ImageNet weights and are adapted to classify the seven FER2013 emotion categories

```python
# Define model creation function

def create_model(model_name):

    if model_name == "ConvNeXt-Tiny":

        weights = models.ConvNeXt_Tiny_Weights.DEFAULT

        model = models.convnext_tiny(weights=weights)

        in_features = model.classifier[2].in_features

        model.classifier[2] = nn.Linear(
            in_features,
            NUM_CLASSES
        )

    elif model_name == "ResNet18":

        weights = models.ResNet18_Weights.DEFAULT

        model = models.resnet18(weights=weights)

        in_features = model.fc.in_features

        model.fc = nn.Linear(
            in_features,
            NUM_CLASSES
        )

    elif model_name == "Swin V2 Tiny":

        weights = models.Swin_V2_T_Weights.DEFAULT

        model = models.swin_v2_t(weights=weights)

        in_features = model.head.in_features

        model.head = nn.Linear(
            in_features,
            NUM_CLASSES
        )

    else:
        raise ValueError("Unknown model name")

    return model.to(DEVICE)


print("Model creation function defined.")
```

```python
# Test model architecture creation

model_names = [
    "ConvNeXt-Tiny",
    "ResNet18",
    "Swin V2 Tiny"
]

for name in model_names:

    model = create_model(name)

    total_parameters = sum(
        p.numel() for p in model.parameters()
    )

    print(
        name,
        "| Total parameters:",
        f"{total_parameters:,}"
    )

    del model

    if torch.cuda.is_available():
        torch.cuda.empty_cache()
```

## **Model Training Strategy**
The models are trained using the following configuration:
Parameter	Value

The model with the best validation loss is saved as the best checkpoint. The test dataset is not used for model selection

```python

# Training and evaluation function with mixed precision (AMP)

def run_epoch(model, loader, criterion, optimizer=None, scaler=None):

    training = optimizer is not None

    model.train() if training else model.eval()

    total_loss = 0
    all_predictions = []
    all_labels = []

    use_amp = torch.cuda.is_available()

    for images, labels in tqdm(loader, leave=False):

        images = images.to(DEVICE, non_blocking=True)
        labels = labels.to(DEVICE, non_blocking=True)

        if training:
            optimizer.zero_grad(set_to_none=True)

        with torch.set_grad_enabled(training):

            with torch.amp.autocast(
                device_type="cuda",
                dtype=torch.float16,
                enabled=use_amp
            ):

                outputs = model(images)
                loss = criterion(outputs, labels)

            if training:

                if use_amp and scaler is not None:

                    scaler.scale(loss).backward()

                    scaler.unscale_(optimizer)

                    torch.nn.utils.clip_grad_norm_(
                        model.parameters(),
                        max_norm=1.0
                    )

                    scaler.step(optimizer)
                    scaler.update()

                else:

                    loss.backward()

                    torch.nn.utils.clip_grad_norm_(
                        model.parameters(),
                        max_norm=1.0
                    )

                    optimizer.step()

        total_loss += loss.item() * images.size(0)

        predictions = outputs.argmax(dim=1)

        all_predictions.extend(predictions.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

    average_loss = total_loss / len(loader.dataset)

    accuracy = accuracy_score(all_labels, all_predictions)

    macro_f1 = f1_score(
        all_labels,
        all_predictions,
        average="macro",
        zero_division=0
    )

    return average_loss, accuracy, macro_f1
```

```python

# Define the model training function with automatic Google Drive checkpoints

import os
import json
import copy
import time
import torch

def train_model(model_name):

    print("\nTraining:", model_name)

    model = create_model(model_name)

    criterion = nn.CrossEntropyLoss(
        label_smoothing=0.1
    )

    optimizer = optim.AdamW(
        model.parameters(),
        lr=LEARNING_RATE,
        weight_decay=WEIGHT_DECAY
    )

    scaler = torch.amp.GradScaler(
    "cuda",
    enabled=torch.cuda.is_available()
)

    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer,
        mode="min",
        factor=0.5,
        patience=2
    )

    history = {
        "train_loss": [],
        "val_loss": [],
        "train_accuracy": [],
        "val_accuracy": [],
        "train_f1": [],
        "val_f1": []
    }

    best_val_loss = float("inf")
    best_state = None
    patience_counter = 0
    start_epoch = 0
    previous_training_time = 0

    safe_name = model_name.replace(" ", "_")

    model_path = os.path.join(
        RESULTS_PATH, "models",
        f"{safe_name}_best.pth"
    )

    checkpoint_path = os.path.join(
        RESULTS_PATH, "models",
        f"{safe_name}_checkpoint.pth"
    )

    history_path = os.path.join(
        RESULTS_PATH, "histories",
        f"{safe_name}_history.json"
    )

    # Resume from the latest saved checkpoint if available

    if os.path.exists(checkpoint_path):

        checkpoint = torch.load(
            checkpoint_path,
            map_location=DEVICE,
            weights_only=False
        )

        model.load_state_dict(checkpoint["model_state_dict"])
        optimizer.load_state_dict(checkpoint["optimizer_state_dict"])
        scheduler.load_state_dict(checkpoint["scheduler_state_dict"])
        if "scaler_state_dict" in checkpoint:
            scaler.load_state_dict(checkpoint["scaler_state_dict"])

        history = checkpoint["history"]
        best_val_loss = checkpoint["best_val_loss"]
        best_state = checkpoint["best_state"]
        patience_counter = checkpoint["patience_counter"]
        start_epoch = checkpoint["epoch"] + 1
        previous_training_time = checkpoint["training_time"]

        print("Previous checkpoint found.")
        print("Resuming from epoch:", start_epoch + 1)

    else:
        print("Starting training from the beginning.")

    current_session_start = time.time()

    for epoch in range(start_epoch, EPOCHS):

        print(f"\nEpoch {epoch + 1}/{EPOCHS}")

        train_loss, train_acc, train_f1 = run_epoch(
            model,
            train_loader,
            criterion,
            optimizer,
            scaler
        )

        val_loss, val_acc, val_f1 = run_epoch(
            model,
            val_loader,
            criterion
        )

        scheduler.step(val_loss)

        history["train_loss"].append(train_loss)
        history["val_loss"].append(val_loss)

        history["train_accuracy"].append(train_acc)
        history["val_accuracy"].append(val_acc)

        history["train_f1"].append(train_f1)
        history["val_f1"].append(val_f1)

        print(f"Train Loss: {train_loss:.4f} | Val Loss: {val_loss:.4f}")
        print(f"Train Accuracy: {train_acc:.4f} | Val Accuracy: {val_acc:.4f}")
        print(f"Train Macro F1: {train_f1:.4f} | Val Macro F1: {val_f1:.4f}")

        # Save the best model whenever validation loss improves

        if val_loss < best_val_loss:

            best_val_loss = val_loss

            best_state = copy.deepcopy(model.state_dict())

            patience_counter = 0

            torch.save(best_state, model_path)

            print("Best model saved to Google Drive.")

        else:
            patience_counter += 1

        total_training_time = (
            previous_training_time +
            time.time() - current_session_start
        )

        # Save complete checkpoint after every epoch

        checkpoint = {
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "scheduler_state_dict": scheduler.state_dict(),
            "scaler_state_dict": scaler.state_dict(),
            "best_val_loss": best_val_loss,
            "best_state": best_state,
            "patience_counter": patience_counter,
            "history": history,
            "training_time": total_training_time
        }

        temporary_checkpoint = checkpoint_path + ".tmp"

        torch.save(checkpoint, temporary_checkpoint)
        os.replace(temporary_checkpoint, checkpoint_path)

        # Save training history

        temporary_history = history_path + ".tmp"

        with open(temporary_history, "w") as file:
            json.dump(history, file, indent=4)

        os.replace(temporary_history, history_path)

        print("Checkpoint and training history saved to Google Drive.")

        if patience_counter >= PATIENCE:

            print("Early stopping triggered.")
            break

    # Load the best-performing weights before returning

    if best_state is not None:
        model.load_state_dict(best_state)

    training_time = (
        previous_training_time +
        time.time() - current_session_start
    )

    print(f"\nTraining completed in {training_time / 60:.2f} minutes.")
    print("Best model:", model_path)

    return model, history, training_time


print("Updated training function defined successfully.")
```

# **Part 5 – Model evaluation**

## **Final Model Evaluation**
After training and validation, the best checkpoint of each model is evaluated on the separate test dataset.

The following metrics are used:
- Accuracy: overall proportion of correct predictions.
- Precision: proportion of positive predictions that are correct.
- Recall: proportion of actual class examples correctly identified.
- Macro F1-score: average F1-score across all emotion classes.
- Weighted F1-score: F1-score averaged according to class support.
These metrics provide a broader evaluation than accuracy alone

```python
# Evaluate models on the test dataset

def evaluate_model(model, loader):

    model.eval()

    all_predictions = []
    all_labels = []

    start_time = time.time()

    with torch.no_grad():

        for images, labels in tqdm(loader):

            images = images.to(DEVICE)

            outputs = model(images)

            predictions = outputs.argmax(dim=1)

            all_predictions.extend(
                predictions.cpu().numpy()
            )

            all_labels.extend(
                labels.numpy()
            )

    elapsed_time = time.time() - start_time

    metrics = {

        "Accuracy": accuracy_score(
            all_labels,
            all_predictions
        ),

        "Macro Precision": precision_score(
            all_labels,
            all_predictions,
            average="macro",
            zero_division=0
        ),

        "Macro Recall": recall_score(
            all_labels,
            all_predictions,
            average="macro",
            zero_division=0
        ),

        "Macro F1": f1_score(
            all_labels,
            all_predictions,
            average="macro",
            zero_division=0
        ),

        "Weighted F1": f1_score(
            all_labels,
            all_predictions,
            average="weighted",
            zero_division=0
        )
    }

    return metrics, all_labels, all_predictions, elapsed_time
```

```python
# Define prediction function

def predict_emotion(model, image_path):

    model.eval()

    image = Image.open(
        image_path
    ).convert("RGB")

    input_tensor = evaluation_transform(image)

    input_tensor = input_tensor.unsqueeze(0).to(DEVICE)

    with torch.no_grad():

        outputs = model(input_tensor)

        probabilities = torch.softmax(
            outputs,
            dim=1
        )[0]

        predicted_index = probabilities.argmax().item()

    return (
        CLASS_NAMES[predicted_index],
        probabilities.cpu().numpy()
    )
```
